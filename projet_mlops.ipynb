## Projet de MLOps 

### Objectif

L'équipe de risque analyse le portefeuille de prêts existants pour prévoir les défauts potentiels futurs et estimer la perte attendue. L'objectif principal est de construire un modèle prédictif qui estime la probabilité de défaut pour chaque client en fonction de ses caractéristiques. Des prédictions précises permettront à la banque d'allouer sufDisamment de capital pour couvrir les pertes potentielles maintenant ainsi la stabilité financieres.
